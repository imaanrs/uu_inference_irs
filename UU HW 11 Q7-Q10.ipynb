## Q7:

#### set-up:

```python
import numpy as np
import pandas as pd

df = pd.read_csv('./data/ames_prices.csv')
print(df['Central.Air'].value_counts())
print(df['Central.Air'].isna().sum())
```

#### a)

```python
df['central_air'] = (df['Central.Air'] == 'Y').astype(int)
p_hat = df['central_air'].mean()
print(p_hat)
```

The proportion of homes with central air is ~0.9331.

#### b)

```python
rng = np.random.default_rng(100)
R = 2000

replications = []
for r in range(R):
    df_r = df.sample(frac=1.0, replace=True, random_state=rng)
    p_r = df_r['central_air'].mean()
    replications.append(p_r)
replications = np.array(replications)

print(replications)
```

The seed = 100 and the # of bootstrap replications is 2,000.

#### c)

```python
se = np.std(replications, ddof=1)
print(se)
```

The bootstrap standard error is ~0.0046.

#### d)

```python
minus90 = p_hat - 1.645 * se
plus90 = p_hat + 1.645 * se

minus95 = p_hat - 1.96 * se
plus95 = p_hat + 1.96 * se

minus99 = p_hat - 2.576 * se
plus99 = p_hat + 2.576 * se

print(f'90% CI: [{minus90}, {plus90}]')
print(f'95% CI: [{minus95}, {plus95}]')
print(f'99% CI: [{minus99}, {plus99}]')
```

#### e)

```python
print('Is 0.90 (null value) inside 90% CI?', minus90 <= 0.90 <= plus90)
print('Is 0.90 (null value) inside 95% CI?', minus95 <= 0.90 <= plus95)
print('Is 0.90 (null value) inside 99% CI?', minus99 <= 0.90 <= plus99)
```

We reject the null at the 90% (& 10% significance level) confidence interval, 95% (& 5% significance level) confidence interval, and 99% (& 1% significance level) confidence interval. This is because none of the intervals include the null value of p = 0.90. We are 99% confident that true value of the proportion of homes with central air falls in the range of (~0.92, ~0.94). I used the widest interval for the interpretation because it is harder to be able to reject the null from this level/CI, and because if we can reject the null from the 99% CI, we can do it for 95% & 90% too.

```python
print(df[['area', 'price']].isna().sum()) #count misisng vals
```

```python
df['arcsinh_area'] = np.arcsinh(df['area'])
df['arcsinh_price'] = np.arcsinh(df['price'])

r_hat = df['arcsinh_area'].corr(df['arcsinh_price'])
print(r_hat)
```

The correlation is ~0.72

#### a)

```python
rng = np.random.default_rng(100)
R = 2000

replications = []
for r in range(R):
    df_r = df.sample(frac=1.0, replace=True, random_state=rng)
    corr_r = df_r['arcsinh_area'].corr(df_r['arcsinh_price'])
    replications.append(corr_r)
replications = np.array(replications)

print(replications)
```

The seed = 100 and the # of bootstrap replications is 2,000.

#### b)

```python
se = np.std(replications, ddof=1)
print(se)
```

The bootstrap standard error is ~0.0101.

#### c)

```python
minus90 = r_hat - 1.645 * se
plus90 = r_hat + 1.645 * se

minus95 = r_hat - 1.96 * se
plus95 = r_hat + 1.96 * se

minus99 = r_hat - 2.576 * se
plus99 = r_hat + 2.576 * se

print(f'90% CI: [{minus90}, {plus90}]')
print(f'95% CI: [{minus95}, {plus95}]')
print(f'99% CI: [{minus99}, {plus99}]')
```

#### d)

r (sample correlation) is positive, with moderate strength within each confidence interval (as it lies between ~.70 to ~.75 in the widest interval). So, larger homes tend to sell for more - they may be newer/have more bathrooms so while area likely may affect the price of a home, tehre are other factors that do too (explaining the moderate strength). The range of values that p (rho) could plausibly be in is ~.70 to ~.75. 

We reject the null hypothesis at the 90% (& 10% significance level) confidence interval, 95% (& 5% significance level) confidence interval, and 99% (& 1% significance level) confidence interval. This is because none of the intervals include the null value of p, rho = 0. We are 99% confident that the correlation between house price and house area falls in the range of (~0.697, ~0.794). I used the widest interval for the interpretation because it is harder to be able to reject the null from this level/CI, and because if we can reject the null from the 99% CI, we can do it for 95% & 90% too.

#### e)

Area and price must remain paired during sampling because each row is one house with its own area & price. The correlation measures whetehr bigger houses tend to have higher prices, so we have to keep each house's area next to it's same price to properly measure the correlation. If you resampled price & area seperately then you'd create fake houses and mismtach the real pairings of price and area so all the measurements would not be representative of the real correlation.

```python
mu_hat = df9['Overall Survival (Months)'].mean()
print(mu_hat)
```

#### a)

```python
rng = np.random.default_rng(100)
R = 2000

replications = []
for r in range(R):
    df9_r = df9.sample(frac=1.0, replace=True, random_state=rng)
    mean_r = df9_r['Overall Survival (Months)'].mean()
    replications.append(mean_r)
replications = np.array(replications)

print(replications)
```

The seed = 100 and the # of bootstrap replications is 2,000.

#### b)

```python
se9 = np.std(replications, ddof=1)
print(se9)
```

The bootstrap standard error is ~2.089.

#### c)

```python
minus90 = mu_hat - 1.645 * se9
plus90 = mu_hat + 1.645 * se9

minus95 = mu_hat - 1.96 * se9
plus95 = mu_hat + 1.96 * se9

minus99 = mu_hat - 2.576 * se9
plus99 = mu_hat + 2.576 * se9

print(f'90% CI: [{minus90}, {plus90}]') #months unit
print(f'95% CI: [{minus95}, {plus95}]')
print(f'99% CI: [{minus99}, {plus99}]')
```

#### d)

```python
print('Is 120 months (null val) inside 90% CI?', minus90 <= 120 <= plus90)
print('Is 120 months (null val) inside 95% CI?', minus95 <= 120 <= plus95)
print('Is 120 months (null val) inside 99% CI?', minus99 <= 120 <= plus99)
```

We reject the null hypothesis at the 90% (& 10% significance level) confidence interval, 95% (& 5% significance level) confidence interval, and 99% (& 1% significance level) confidence interval. This is because none of the intervals include the null value of mu = 120 months. We are 99% confident that the mean recorded time to death/last follow-up falls in the range of (~123.71, ~134.48) months. I used the widest interval for the interpretation because it is harder to be able to reject the null from this level/CI, and because if we can reject the null from the 99% CI, we can do it for 95% & 90% too.
